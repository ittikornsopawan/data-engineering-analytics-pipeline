```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

matches_path = "../data/matches_1930_2022.csv"

df = pd.read_csv(matches_path)
```

```python
def calculation_goal_differences(match):
    if match["home_score"] > match["away_score"]:
        return match["home_score"] - match["away_score"]
    elif match["home_score"] < match["away_score"]:
        return match["away_score"] - match["home_score"]
    else:
        return 0
```

```python
def match_winner(match):
    if match["home_score"] > match["away_score"]:
        return match["home_team"]
    elif match["home_score"] < match["away_score"]:
        return match["away_team"]
    else:
        return "DRAW"
```

```python
def check_draw(match):
    if match["winner"] == "DRAW":
        return True
    else:
        return False
```

```python
def check_high_scoring_match(match):
    if(match["total_goals"] > 5):
        return True
    else:
        return False
```

```python
matches = df[["home_team", "away_team", "Year", "home_score", "away_score"]].copy()

matches = matches.rename(columns = {
    "away_team": "away_team",
    "home_team": "home_team",
    "Year": "year",
    "away_score": "away_score",
    "home_score": "home_score"
})

matches["total_goals"] = (
    matches["home_score"] + matches["away_score"]
)

matches["goal_difference"] = matches.apply(calculation_goal_differences, axis=1)
matches["winner"] = matches.apply(match_winner, axis=1)
matches["is_draw"] = matches.apply(check_draw, axis=1)
matches["is_high_scoring_match"] = matches.apply(check_high_scoring_match, axis=1)

matches
```
