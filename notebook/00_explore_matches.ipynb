```python
matches_by_year = df["Year"].value_counts().sort_index()

matches_by_year
```

```python
matches_by_year.plot(kind="bar", figsize=(12, 5))

plt.title("Number of Matches by World Cup Year")

plt.xlabel("Year")

plt.ylabel("Number of Matches")

plt.show()
```

```python
all_teams = pd.concat([df["home_team"], df["away_team"]]).dropna().unique()

print("Total teams:", len(all_teams))

sorted(all_teams)
```

```python
missing_values = df.isna().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

missing_values
```

```python


missing_percent = (df.isna().sum() / len(df)) * 100
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

missing_percent
```
