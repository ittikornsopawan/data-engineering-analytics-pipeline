```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

matches_path = "../data/matches_1930_2022.csv"

df = pd.read_csv(matches_path)
```

```python
matches = df[["home_team", "away_team", "Year", "home_score", "away_score"]].copy()

matches = matches.rename(columns = {
    "away_team": "away_team",
    "home_team": "home_team",
    "Year": "year",
    "away_score": "away_score",
    "home_score": "home_score"
})

matches["goals"] = (
    matches["home_score"] + matches["away_score"]
)

matches
```

```python
total_match = (
    matches
    .groupby(["year"])
    .size()
    .reset_index(name="total_matches")
)

total_match
```

```python
home_teams = matches[["year", "home_team", "home_score"]].copy()
home_teams = home_teams.rename(columns={
    "home_team": "team",
    "home_score": "goals"
})

away_teams = matches[["year", "away_team", "away_score"]].copy()
away_teams = away_teams.rename(columns={
    "away_team": "team",
    "away_score": "goals"
})

teams_by_year = pd.concat([home_teams, away_teams], ignore_index=True)

teams_by_year
```

```python
total_teams = (
    teams_by_year
    .groupby("year")["team"]
    .nunique()
    .reset_index(name="total_teams")
)

total_teams
```

```python
year_summary = total_match.merge(
    total_teams,
    on="year",
    how="left"
)

year_summary
```

```python
total_goals = (
    teams_by_year
    .groupby("year")["goals"]
    .sum()
    .reset_index(name="total_goals")
)

total_goals
```

```python
year_summary = year_summary.merge(
    total_goals,
    on="year",
    how="left"
)

year_summary
```

```python
year_summary["avg_goals_per_match"] = (
    year_summary["total_goals"] / year_summary["total_matches"]
)

year_summary
```

```python
max_goals_in_match = (
    matches.groupby("year")["goals"]
    .max()
    .reset_index(name="max_goals_in_match")
)

max_goals_in_match
```

```python
year_summary = year_summary.merge(
    max_goals_in_match,
    on="year",
    how="left"
)

year_summary
```

```python
high_score = matches[matches["goals"] > 5 ]

high_scoring_matches = (
    high_score.groupby("year")["goals"]
    .count()
    .reset_index(name="high_scoring_matches")
)

high_scoring_matches
```

```python
year_summary = year_summary.merge(
    high_scoring_matches,
    on="year",
    how="left"
)

year_summary
```
